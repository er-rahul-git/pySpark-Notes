# Dataframe wrangling
<hr style = "border:2px solid black" ></hr>

```python
spark
```

```python
# read the dataset
# if you do not use inferSchema=True it will assign each columns the type string
df_pyspark = spark.read.option('header', 'true').csv(
    '/content/sample_data/california_housing_test.csv', inferSchema=True)
```

```python
# Check the schema
df_pyspark.printSchema()
```

```python
df_pyspark = spark.read.csv('/content/sample_data/california_housing_test.csv', header=True, inferSchema=True)
df_pyspark.show()
```

```python
# Check the schema
df_pyspark.printSchema()
```

```python
type(df_pyspark)
```

```python
df_pyspark.head(3)
```

```python
df_pyspark.show()
```

```python
df_pyspark.select(['Name', 'Experience']).show()
```

```python
df_pyspark['Name']
```

```python
df_pyspark.dtypes
```

```python
df_pyspark.describe().show()
```

```python
# Adding Columns in data frame
df_pyspark = df_pyspark.withColumn(
    'Experience After 2 year', df_pyspark['Experience']+2)
```

```python
df_pyspark.show()
```

```python
# Drop the columns
df_pyspark = df_pyspark.drop('Experience After 2 year')
```

```python
df_pyspark.show()
```

```python
# Rename the columns
df_pyspark.withColumnRenamed('Name', 'New Name').show()
```
